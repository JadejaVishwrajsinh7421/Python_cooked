### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius

r = float(input("Enter radius: "))
c = Circle(r)

print("Area:", c.area())
print("Perimeter:", c.perimeter())
```

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Balance:", self.balance)

balance = float(input("Enter initial balance: "))
acc = BankAccount(balance)

amount = float(input("Enter deposit amount: "))
acc.deposit(amount)
acc.check_balance()

amount = float(input("Enter withdraw amount: "))
acc.withdraw(amount)
acc.check_balance()
```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class Inventory:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def add(self, qty):
        self.quantity += qty

    def remove(self, qty):
        if qty <= self.quantity:
            self.quantity -= qty
        else:
            print("Not enough stock")

    def update(self, price):
        self.price = price

    def display(self):
        print("Item:", self.name)
        print("Price:", self.price)
        print("Quantity:", self.quantity)

name = input("Enter item name: ")
price = float(input("Enter price: "))
quantity = int(input("Enter quantity: "))

item = Inventory(name, price, quantity)
item.display()

qty = int(input("Enter quantity to add: "))
item.add(qty)
item.display()

qty = int(input("Enter quantity to remove: "))
item.remove(qty)
item.display()

new_price = float(input("Enter new price: "))
item.update(new_price)
item.display()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time:
    def __init__(self,hour,minute):
        self.hour = hour
        self.minute = minute
    
    def add(self,t):
        total_m = self.minute + t.minute
        extra_hr = total_m // 60
        final_min = total_m % 60
        final_hr = self.hour + t.hour + extra_hr
        return Time(final_hr, final_min)

t1 = Time(2, 50)
t2 = Time(1, 30)

t3 = t1.add(t2)
print(t3.hour, "hour", t3.minute, "minute")
```
