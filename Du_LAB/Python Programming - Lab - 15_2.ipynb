### 10) Calculate area of a ractangle using object as an argument to a method.

### 12) Calculate the area of a rectangle. 
### Include a Constructor, a method to calculate area named area() and a method named output() that prints the output and is invoked by area().
### Also define a class method that compares the two sides of reactangle. An object is instantiated only if the two sides are different; otherwise a message should be displayed : THIS IS SQUARE.

```python
class Rectangle:
    def __init__(self, l, b):
        self.length = l
        self.breath = b

    @classmethod
    def area(cls, obj):
        if obj.length != obj.breath:
           result = obj.length * obj.breath
           obj.output(result)
        else:
            print("THIS IS SQUARE")

    
    def output(self, result):
        print(f"Area of rectangle: {result}")

r2 = Rectangle(5,5)
Rectangle.area(r2)
```

### 14) Create a class Profit that has a method named getProfit that accepts profit from the user.
### Create a class Loss that has a method named getLoss that accepts loss from the user.
### Create a class BalanceSheet that inherits from both classes Profit and Loss and calculates the balanace. It has two methods getBalance() and printBalance().

```python
class Profit:
    def __init__(self):
        self.profit = 0
    
    def get_profit(self):
        self.profit = int(input("Enter profit amount: "))

class loss:
    def __init__(self):
        self.loss = 0
    
    def get_loss(self):
        self.loss = int(input("Enter loss amount: "))

class BalanceSheet(Profit, loss):
    def getBalance(self):
        self.balance = self.profit - self.loss

    def printBalance(self):
        print(f"Balance is: {self.balance}")

b1 = BalanceSheet()
b1.get_profit()
b1.get_loss()
b1.getBalance()
b1.printBalance()
```

### 15) WAP to demonstrate all types of inheritance.

```python
# 1. Single Inheritance
class A:
    def show_A(self):
        print("Class A")

class B(A):
    def show_B(self):
        print("Class B (Single Inheritance)")


# 2. Multiple Inheritance
class C:
    def show_C(self):
        print("Class C")

class D:
    def show_D(self):
        print("Class D")

class E(C, D):
    def show_E(self):
        print("Class E (Multiple Inheritance)")


# 3. Multilevel Inheritance
class F:
    def show_F(self):
        print("Class F")

class G(F):
    def show_G(self):
        print("Class G")

class H(G):
    def show_H(self):
        print("Class H (Multilevel Inheritance)")


# 4. Hierarchical Inheritance
class I:
    def show_I(self):
        print("Class I")

class J(I):
    def show_J(self):
        print("Class J")

class K(I):
    def show_K(self):
        print("Class K (Hierarchical Inheritance)")


# 5. Hybrid Inheritance (Combination)
class L:
    def show_L(self):
        print("Class L")

class M(L):
    def show_M(self):
        print("Class M")

class N(L):
    def show_N(self):
        print("Class N")

class O(M, N):
    def show_O(self):
        print("Class O (Hybrid Inheritance)")

print("\nSingle Inheritance:")
b = B()
b.show_A()
b.show_B()

print("\nMultiple Inheritance:")
e = E()
e.show_C()
e.show_D()
e.show_E()

print("\nMultilevel Inheritance:")
h = H()
h.show_F()
h.show_G()
h.show_H()

print("\nHierarchical Inheritance:")
j = J()
k = K()
j.show_I()
j.show_J()
k.show_I()
k.show_K()

print("\nHybrid Inheritance:")
o = O()
o.show_L()
o.show_M()
o.show_N()
o.show_O()
```

### 16) Create a Person class with a constructor that takes two arguments name and age. 
### Create a child class Employee that inherits from Person and adds a new attribute salary. 
### Override the __init__ method in Employee to call the parent class's __init__ method using the super() and then initialize the salary attribute.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def show(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

    def show(self):
        print(f"Salary: {self.salary}")

e1 = Employee("Rakshit", 20, 100000)    
e1.show()
```

### 17) Create a Shape class with a draw method that is not implemented. 
### Create three child classes Rectangle, Circle, and Triangle that implement the draw method with their respective drawing behaviors. 
### Create a list of Shape objects that includes one instance of each child class, and then iterate through the list and call the draw method on each object.

```python
# Base class
class Shape:
    def draw(self):
        raise NotImplementedError("Subclass must implement draw method")


# Child class: Rectangle
class Rectangle(Shape):
    def draw(self):
        print("Drawing Rectangle")


# Child class: Circle
class Circle(Shape):
    def draw(self):
        print("Drawing Circle")


# Child class: Triangle
class Triangle(Shape):
    def draw(self):
        print("Drawing Triangle")


# Create list of Shape objects
shapes = [Rectangle(), Circle(), Triangle()]

# Iterate and call draw()
for shape in shapes:
    shape.draw()
```
