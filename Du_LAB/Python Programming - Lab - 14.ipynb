### 06) WAP to create bar chart showing the marks obtained by students in different subjects.
### The program should also display the marks on the top of the respective bar.

```python
Subjects = ["Python", "DBMS", "SE", "PS", "PE-1"]
Marks = [78, 85, 72, 80, 90]
```

```python
import matplotlib.pyplot as plt

Subjects = ["Python", "DBMS", "SE", "PS", "PE-1"]
Marks = [78, 85, 72, 80, 90]

plt.bar(Subjects, Marks)

for i in range(len(Subjects)):
    plt.text(i, Marks[i], Marks[i], ha='center')

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks in Different Subjects")

plt.show()
```

### 07) WAP to create a horizontal bar chart representing the number of books sold in different categories. 
### The program should also display the number of books sold at the end of the respective bar.

```python
Categories = ["Fiction", "Science", "History", "Technology", "Biography"]
Books Sold = [120, 90, 70, 110, 60]
```

```python
import matplotlib.pyplot as plt

Categories = ["Fiction", "Science", "History", "Technology", "Biography"]
Books_Sold = [120, 90, 70, 110, 60]

plt.barh(Categories, Books_Sold)

for i in range(len(Categories)):
    plt.text(Books_Sold[i], i, Books_Sold[i], va='center')

plt.xlabel("Books Sold")
plt.ylabel("Categories")
plt.title("Books Sold in Different Categories")

plt.show()
```

### 08) WAP to create a scatter plot representing the relationship between advertising expenditure and sales revenue of a company.

```python
Advertising Cost (₹ thousands) = [5, 8, 10, 12, 15, 18, 20]
Sales Revenue (₹ thousands) = [50, 60, 65, 70, 80, 90, 95]
```

```python
import matplotlib.pyplot as plt

Advertising_Cost = [5, 8, 10, 12, 15, 18, 20]
Sales_Revenue = [50, 60, 65, 70, 80, 90, 95]

plt.scatter(Advertising_Cost, Sales_Revenue)

for i in range(len(Advertising_Cost)):
    plt.text(Advertising_Cost[i], Sales_Revenue[i], f"({Advertising_Cost[i]}, {Sales_Revenue[i]})")

plt.xlabel("Advertising Cost (₹ thousands)")
plt.ylabel("Sales Revenue (₹ thousands)")
plt.title("Advertising vs Sales Revenue")

plt.show()
```

### 09) WAP to create a histogram representing the distribution of marks obtained by students in a class.
