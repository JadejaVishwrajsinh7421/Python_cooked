### 01) WAP to find their union, intersection, and difference of the given two sets.
**Sample Input:**
```
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

```
**Sample Output:**
```
Union: {1, 2, 3, 4, 5, 6}
Intersection: {3, 4}
Difference (A-B): {1, 2}

```

````python
n1 = int(input("Enter number of elements in set A: "))
A = []
for i in range(n1):
    A.append(int(input()))

n2 = int(input("Enter number of elements in set B: "))
B = []
for i in range(n2):
    B.append(int(input()))

setA = set(A)
setB = set(B)

print("Union:", setA | setB)
print("Intersection:", setA & setB)
print("Difference (A-B):", setA - setB)
````

### 05) WAP to Merge two dictionaries. If a key exists in both dictionaries, add their values.

**Sample Input:**
```
Dict1 = {1: 10, 2: 20, 3: 30}
Dict2 = {2: 40, 3: 50, 4: 60}

```
**Sample Output:**
```
{1: 10, 2: 60, 3: 80, 4: 60}
```

````python
n1 = int(input("Enter number of key-value pairs in Dict1: "))
dict1 = {}
for i in range(n1):
    key = int(input("Enter key for Dict1: "))
    value = int(input("Enter value for Dict1: "))
    dict1[key] = value

n2 = int(input("Enter number of key-value pairs in Dict2: "))
dict2 = {}
for i in range(n2):
    key = int(input("Enter key for Dict2: "))
    value = int(input("Enter value for Dict2: "))
    dict2[key] = value

for key in dict2:
    if key in dict1:
        dict1[key] = dict1[key] + dict2[key]
    else:
        dict1[key] = dict2[key]

print(dict1)
````

### 06) WAP to Sort a dictionary in ascending order of its values.
Hint: in sorted(), use udf for key parameter


**Sample Input:**
```
{'Math': 70, 'Physics': 85, 'Chemistry': 60}

```
**Sample Output:**
```
{'Chemistry': 60, 'Math': 70, 'Physics': 85}
```

````python
n = int(input("Enter number of key-value pairs: "))
d = {}
for i in range(n):
    key = input("Enter key: ")
    value = int(input("Enter value: "))
    d[key] = value

def get_value(item):   # used to extract the dictionary value so sorting happens by values instead of keys.
    return item[1]

items = list(d.items())   # produces (key, value) pairs.
items.sort(key=get_value) # before sorting each item, pass it to the function get_value and use the returned value for sorting.

sorted_dict = {}
for k, v in items:
    sorted_dict[k] = v

print(sorted_dict)
````

### 07) WAP to find the keys having maximum and minimum values in a dictionary.

**Sample Input:**
```
{'A': 50, 'B': 90, 'C': 30}
```
**Sample Output:**
```
Max Value Key: B
Min Value Key: C

```

````python
n = int(input("Enter number of key-value pairs: "))
d = {}
for i in range(n):
    key = input("Enter key: ")
    value = int(input("Enter value: "))
    d[key] = value

max_value = max(d.values())
min_value = min(d.values())

for k in d:
    if d[k] == max_value:
        print("Max Value Key:", k)
        break

for k in d:
    if d[k] == min_value:
        print("Min Value Key:", k)
        break
````

### 10) WAP to swap keys and values of a dictionary.
**Sample Input:**
```
{1: 'A', 2: 'B', 3: 'C'}
```

```
{'A': 1, 'B': 2, 'C': 3}
```

### 11) WAP to find the student with the highest total marks, given student names and their marks.

**Sample Input:**
```
{'Amit': [70, 80, 90], 'Riya': [85, 75, 80], 'Neha': [60, 70, 65]}
```
**Sample Output:**
```
Topper: Amit
```

````python
n = int(input("Enter number of students: "))
m = int(input("Enter number of subjects: "))

students = {}

for i in range(n):
    name = input("Enter student name: ")
    marks = []
    for j in range(m):
        marks.append(int(input("Enter mark: ")))
    students[name] = marks

topper = None
highest_total = None

for name in students:
    total = 0
    for mark in students[name]:
        total = total + mark
    if highest_total is None or total > highest_total:
        highest_total = total
        topper = name

print("Topper:", topper)
````

### 12) WAP to extract keys whose values are even numbers.

**Sample Input:**
```
{1: 11, 2: 20, 3: 15, 4: 40}
```
**Sample Output:**
```
[2, 4]
```

````python
n = int(input("Enter number of key-value pairs: "))
d = {}

for i in range(n):
    key = int(input("Enter key: "))
    value = int(input("Enter value: "))
    d[key] = value

result = []

for k in d:
    if d[k] % 2 == 0:
        result.append(k)

print(result)
````

### 15) WAP to create a dictionary using dictionary comprehension from two lists containing student names and marks. Include only those students in the dictionary whose marks are greater than or equal to 75.

**Sample Input:**
```
students = ['Amit', 'Riya', 'Neha', 'Karan']
marks = [78, 72, 85, 60]
```
**Sample Output:**
```
{'Amit': 78, 'Neha': 85}
```

````python
n = int(input("Enter number of students: "))

students = []
marks = []

for i in range(n):
    students.append(input("Enter student name: "))

for i in range(n):
    marks.append(int(input("Enter marks: ")))

result = {students[i]: marks[i] for i in range(n) if marks[i] >= 75}

print(result)
````
