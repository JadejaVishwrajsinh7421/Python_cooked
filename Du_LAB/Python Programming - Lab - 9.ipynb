### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def is_prime(n):
    if n <= 1:
        return 0
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return 0
    return 1

n = int(input("Enter a number: "))

print(is_prime(n))
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def primes_between(start, end):
    primes = []
    for num in range(max(2, start), end + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return primes


start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

print("Prime numbers between", start, "and", end, "are:")
print(primes_between(start, end))
```

### 07) Write a function to calculate the sum of the first element of each tuples inside the list.

```python
def sum_first_elements(tuples_list):
    return sum(t[0] for t in tuples_list)

n = int(input())
tuples_list = [tuple(map(int, input().split())) for _ in range(n)]
print(sum_first_elements(tuples_list))
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input())
print(fibonacci(n))
```

### 09) Write a function to get the name of the student based on the given rollno.
#### Example: Given dict1 = {101:'Ajay', 102:'Rahul', 103:'Jay', 104:'Pooja'} find name of student whose rollno = 103

```python
def get_student_name(d, rollno):
    return d.get(rollno)

n = int(input())
d = {}
for _ in range(n):
    k, v = input().split()
    d[int(k)] = v
rollno = int(input())
print(get_student_name(d, rollno))
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
def scoress(scores):
    return sum(x for x in scores if x % 10 == 0)

scores = list(map(int, input().split()))
print(scoress(scores))
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def invert_dict(d):
    return {v: k for k, v in d.items()}

print(invert_dict({'a': 10, 'b': 20, 'c': 30, 'd': 40}))
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def count_up_low(s):
    up = sum(1 for ch in s if ch.isupper())
    low = sum(1 for ch in s if ch.islower())
    return up, low

print(count_up_low("AbcDEfgh"))
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
# Positional Args
def pos(a, b):
    return a + b

print(pos(5, 10))


# Keyword Args
def kw(a=0, b=0):
    return a + b

print(kw(b=5, a=3))


# Default Args
def default(a=10, b=20):
    return a + b

print(default())


# Variable length 
def var(*args, **kwargs):
    print(args, kwargs)

var(1,2,3, x=10, y=20)


# Keyword-only & Positional-only
def mix(a, /, b, *, c):
    print(a, b, c)

mix(1, b=2, c=3)
```
