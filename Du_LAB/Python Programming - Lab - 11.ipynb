### Practical 1: Count Lines, Words and Characters

**Problem Definition:**  
Write a Python program to read text from a file and count the total number of lines, words, and characters. The result should be written to an output file.


```python
file1 = input("Enter input file name: ")
file2 = input("Enter output file name: ")

with open(file1, "r") as f:
    text = f.read()

lines = text.count("\n") + 1
words = len(text.split())
characters = len(text)

with open(file2, "w") as f:
    f.write(f"Lines: {lines}\n")
    f.write(f"Words: {words}\n")
    f.write(f"Characters: {characters}\n")

print("Count written to output file successfully!")
```


### Practical 2: Search a Word in a File

**Problem Definition:**  
Write a Python program to search a given word in a text file and count its occurrences. Store the result in an output file.


```python
file1 = input("Enter input file name: ")
file3 = input("Enter output file name: ")
word = input("Enter word to search: ")

with open(file1, "r") as f:
    text = f.read()

count = text.split().count(word)

with open(file3, "w") as f:
    f.write(f"Word: {word}\n")
    f.write(f"Occurrences: {count}\n")

print("Result written to output file successfully!")
```


### Practical 3: Student Marks and Average

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Store marks in a list, calculate the average using a user-defined function, and write student details along with average marks to an output file.


```python
file4 = input("Enter input file name: ")
file5 = input("Enter output file name: ")

def average(marks):
    return sum(marks) / len(marks)

with open(file4, "r") as f:
    lines = f.readlines()

marks = []
students = []

for line in lines:
    roll, name, mark = line.split()
    mark = int(mark)
    marks.append(mark)
    students.append((roll, name, mark))

avg = average(marks)

with open(file5, "w") as f:
    for roll, name, mark in students:
        f.write(f"{roll} {name} {mark}\n")
    f.write(f"Average Marks: {avg}\n")

print("Student details and average written successfully!")
```


### Practical 4: Maximum and Minimum using Tuple

**Problem Definition:**  
Write a Python program to read numbers from a file, store them in a tuple, and find the maximum and minimum values. Write the result to an output file.


```python
file6 = input("Enter input file name: ")
file7 = input("Enter output file name: ")

with open(file6, "r") as f:
    numbers = tuple(map(int, f.read().split()))

maximum = max(numbers)
minimum = min(numbers)

with open(file7, "w") as f:
    f.write(f"Numbers: {numbers}\n")
    f.write(f"Maximum: {maximum}\n")
    f.write(f"Minimum: {minimum}\n")

print("Maximum and minimum written successfully!")
```


### Practical 5: Remove Duplicate Words using Set

**Problem Definition:**  
Write a Python program to read words from a file and remove duplicate words using a set. Store the unique words in an output file.


```python
file1 = input("Enter input file name: ")
file8 = input("Enter output file name: ")

with open(file1, "r") as f:
    words = f.read().split()

unique_words = set(words)

with open(file8, "w") as f:
    for word in unique_words:
        f.write(word + "\n")

print("Duplicate words removed successfully!")
```


### Practical 6: Word Frequency using Dictionary

**Problem Definition:**  
Write a Python program to count the frequency of each word in a text file using a dictionary and write the result to an output file.


```python
file1 = input("Enter input file name: ")
file9 = input("Enter output file name: ")

with open(file1, "r") as f:
    words = f.read().split()

freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

with open(file9, "w") as f:
    for word, count in freq.items():
        f.write(f"{word}: {count}\n")

print("Word frequency written successfully!")
```


### Practical 7: Student Result and Grade Calculation

**Problem Definition:**  
Write a Python program to read roll number, name, and marks of students from a file. Use a user-defined function to determine grade (A/B/C/Fail) and write the student result to an output file.

**Grade Criteria:**  
- Marks ≥ 75 : Grade A  
- Marks ≥ 60 : Grade B  
- Marks ≥ 40 : Grade C  
- Marks < 40 : Fail


```python
file4 = input("Enter input file name: ")
file10 = input("Enter output file name: ")

def grade(mark):
    if mark >= 75:
        return "A"
    elif mark >= 60:
        return "B"
    elif mark >= 40:
        return "C"
    else:
        return "Fail"

with open(file4, "r") as f:
    lines = f.readlines()

with open(file10, "w") as f:
    for line in lines:
        if not line.strip():
            continue
        roll, name, mark = line.split()
        mark = int(mark)
        g = grade(mark)
        f.write(f"{roll} {name} {mark} {g}\n")

print("Student results written successfully!")
```
