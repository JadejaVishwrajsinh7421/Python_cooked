# Arithmeatic operation

```python
import  numpy as np
```

```python
arr1 =  np.array([12,23,45,5006,776,79,21])
```

```python
# min value
min =  np.min(arr1)
print("min",min)

# max value
max = np.max(arr1)
print("max",max)
```

```python
# for min element's argument
x =  np.argmin(arr1)
print(x)

# simalar for max element's index
y = np.argmax(arr1)
print(y)
```

```python
# for n- array we can chose min and max acc to rows and col

arr2 =  np.array([[1,34,564,14],[456,8,6,100]])
#for  axis =0 => acc to row
x =  np.min(arr2,axis = 0) 
print(x)

x1 =  np.max(arr2,axis =0)
print(x1)

# for  axis =1 => acc to col
y =  np.min(arr2,axis = 1) 
print(y)

y1  = np.max(arr2,axis = 1)
print(y1)
```

```python
# trigo functions

sin = np.sin(arr2)
print(sin)
print()
cos = np.cos(arr2)
print(cos)
print()
tan = np.tan(arr2)
print(tan)
print()
cot = np.reciprocal(tan)
print(cot)
print()
sec = np.reciprocal(np.cos(arr2))
print(sec)
print()
cosec = np.reciprocal(np.sin(arr2))
print(cosec)
```

```python
# cumlative sum of the elements of array
csum = np.cumsum(arr2)
print(arr2)
print(csum) 
```

```python
# square root func
print(np.sqrt(arr1))
print(np.sqrt(arr2))
```
