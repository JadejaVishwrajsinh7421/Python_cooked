# Arithmeatic operation

```python
import  numpy as np
```

```python
x =  np.array([12,23,45,56,776,79])
```

ADD

```python
xadd =  x+3 
# for single number 
print(xadd)
```

```python
# addition of 1- d array
var1 = np.array([463,7896,78,63,94,65])
var1 =  np.add(var1,x)
print(var1)
```

SUB

```python
xsub=  x-3 
# for single number 
print(xsub)

# subtrcation of 1- d array
var1 = np.array([463,7896,78,63,94,65])
var2 =  np.subtract(var1,x)
print(var2)
```

DIVIDE

```python
xdiv =  x/3 
# for single number 
print(xdiv)

# divide of 1- d array
var1 = np.array([463,7896,78,63,94,65])
new =  np.divide(var1,x)
print(new)
```

Mod

```python
xmod =  x%3 
# for single number 
print(xmod)

# modulus of 1- d array
var1 = np.array([463,7896,78,63,94,65])
new =  np.mod(var1,x)
print(new)
```

Multiply

```python
xmul =  x*3 
# for single number 
print(xmul)

# Multiply of 1- d array
var1 = np.array([463,7896,78,63,94,65])
new =  np.multiply(var1,x)
print(new)
```

Reciprocal 

```python
# Multiply of 1- d array
var1 = np.array([4.63,78.96,0.78,0.63,92.4,45.65])
new =  np.reciprocal (var1)
print(new)
```

# For 2-d Array

```python
var2 = np.array([[1,23],[4578,63]])
var3 = np.array([[1,2],[1,6]])

# For addition
new =  np.add(var2,var3)
print(new)
# it follows the matrix addition rule in this...


# For Multiplication
new =  np.multiply(var2,var3)
print(new)
# it  does not follows the matrix Multiplication rule in this...

# for actual matrix dot product (two mnatrix Multiplication )
new =  var2 @ var3
print(new)
```
