# Random number

```python
import numpy as np
```

1 Rand()

```python
ran = np.random.rand(4)
print(ran)
```

```python
# for 2-d arrays
ran = np.random.rand(2,5)
print(ran)
```

2 Randn()

```python
ran2 = np.random.randn(5)
print(ran2)
```

```python
ran3 = np.random.randn(5,7,3)
print(ran3)
```

 3 Randf()

```python
import numpy as np
ra3 = np.random.ranf(7)
print(ra3)

print()
ra3 = np.random.ranf((2,3))
print(ra3)
```
