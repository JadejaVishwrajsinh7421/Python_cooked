Data conversion in array:

```python
# method  1
x =  np.array([1,2,34,554,67543,12])
print("Data type:" ,x.dtype)
print(x)
# i want to change in data type int8 by this way
x =  np.array([1,2,34,554,67543,12],dtype = np.float64)
print("Data type:" ,x.dtype)
print(x)
```

```python
#method -2  by func way():

x =  np.array([1,2,34,554,67543,12])

new  = np.float32(x)
print("Data type old:" ,x.dtype)
print("Data type new:" ,new.dtype)

print(x)
print(new)
```
