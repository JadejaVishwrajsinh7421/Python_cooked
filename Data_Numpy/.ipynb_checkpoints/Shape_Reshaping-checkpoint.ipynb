# Reshape of Array

```python
arr2  = np.array([1,2,45,78,9,5,12,21,43,45,6,7])
print(arr2)
print(arr2.ndim)
print()

new  =  arr2.reshape(2,3,2) 
# we check new dimens 
print(new.ndim)
# 2 row and  3col elements arrange
print(new)
```

```python
#  again going back to n-d to 1-d array back
new1  = new.reshape(-1)
print(new1)
print(new1.ndim)
```
