# Iterating  array

```python
import numpy as np
```

```python
var1 = np.array([1,2,33,56,621,34])
print(var1)
print()
for i in var1:
    print(i)
```

```python
var2 = np.array([[1,34,54],[243,754,875]])
# for j in var2:
#     print(j)

for k  in var2:
    for l in k:
        print(l)

# same for 3-d we have to add 3  for loops

var3 = np.array([[[1,2],[21,67],[3,4]]])
print(var3)

for k in var3:
    for l in k:
        for i in l:
            print(i)
```

```python
#  for withou use of the loop we use the function directly
var2 = np.array([[1,34,54],[243,754,875]])

for i in np.nditer(var2,flags = ['buffered'],op_dtypes=["S"]):
    print(i)
```

```python
for i,d in np.ndenumerate(var2):
    print(i,d)
```
