# Copy and view functions

```python
import numpy as np
```

```python
var1  = np.array([1,3,34,54])
cop1 = var1.copy()

print(var1)
print(cop1)

# afeect the original 
var1[0] =78
print(var1)
print(cop1)
```

```python
x = np.array([897,456,75,51])
vie1  = x.view()

print(x)
print(vie1)

x[0] =78
print(x)
print(vie1)
```
