```python
# Creating array with diifrent need and format
```

1 : Zeroes Array

```python
import numpy as np
arr_zero = np.zeros(4)
print(arr_zero)
# <!-- it gives  1-D array -->
print()
#for 2-d array
ar_zero2 = np.zeros((3,4))
# <!-- (Row,Col) in the way -->
print(ar_zero2)
```

2 : One array

```python
arr_one_1  = np.ones(5)
print(arr_one_1)
print()
arr_one_2 = np.ones((2,5))
print(arr_one_2)
```

3 : Empty Array

```python
arr_emp_1 = np.empty(3)
print(arr_emp_1)
#it gives data of old array version or diffrent ans may be 
```

4 : Range Array

```python
arr_ran = np.arange(4,7,2)
print(arr_ran)
# similar to range function funcnality
# start with 0 end upto n-1 

#it will only generate 1-d array 
#  for 2-d array
# we use reshape()
arr_ran_2 = np.arange(12).reshape(4,3)
print(arr_ran_2)

# in we reshape we only give factors of theww size of the array in arange func()
```

5 : Diagonal element(1) (Identity matrix)

```python
arr_dia = np.eye(3)
print(arr_dia)
```

```python
arr_dia_rand = np.eye(3,5)
print(arr_dia_rand)
```

6 : spaced interval array linspace()

```python
arr_space = np.linspace(1,10,num=5)
print(arr_space)
```
