# Indexing

```python
import numpy as np
```

```python
var1 = np.array([1,34,57,98])
                #(0,1 ,2 ,3))
                #(-4,-3,-2,-1)
print(var1[3])
print(var1[-2])
```

```python
# for 2-d array
var2 = np.array([[9,8,3],[4,5,6]])

print(var2.ndim)
print()
print(var2[0,1]) # for 8(row,col)

# for 3-d array

var3 = np.array([[[1,2],[1,2]] ,[[3,2],[4,5]]])
print(var3.ndim)
print()
print(var3[0,1,0])#last 2 in 2nd array 
print(var3[1,0,0])#3 from 2nd row array
```

# Silicing

```python
var4 = np.array([12,846,97,46,347,76])
print(var4)
print(var4.ndim)
```

```python
print("846 to 46->",var4[1:4])
print("97 to end->",var4[2:])
print("12 to 46->",var4[:4])
print("one one jump ->",var4[::1])
print("one one jump ->",var4[::-1]) #reverse the array
print("some patteren ->",var4[2:6:2])
print(var4[-3:-1])
print("-ve slicing->",var4[-4:3])
```

```python
var5 = np.array([[45,79,43,21],[134,94,35,309],[54,79,5,6]])

print(var5.ndim)
```

```python
print("2nd row 94 to 309->",var5[1,1:4])
print(var5[0:3,2]) #(all 3 row return index 2 from all )
print(var5[0:3:2, 1:4])
```
