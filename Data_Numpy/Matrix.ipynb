# Matrix

```python
import numpy as np
```

```python
var1 = np.matrix([[1,2,3],[2,45,67]])

print(var1)
print(var1.ndim)
print(type(var1))
```

```python
var2 = np.array([[1,2,3],[2,45,67]])
print(var2)
print(var2.ndim)
print(type(var2))
```

```python
var2 = np.array([[1,2,3],[2,45,67],[79,16,63]])
var3  = np.array([[2,3,4],[6,7,8],[16,67,34]])

# add
print(var2+var3)

# sub
print(var2-var3)

# mul by matrix rule
print(var2 @ var3)
print(var2.dot(var3))
print(var2 * var3) #it is diffrent 

#div
print(var2 / var3)
print(var3 /var2)
```

# Inbulit functions of matrix

```python
# transpose matrix 

var4 = np.matrix([[1,23,34],[65,97,65]])
print(var4)

print(np.transpose(var4))
print(var4.T) #same work
```

```python
# swap axes 
print(np.swapaxes(var4,0,1))

var5 = np.matrix([[1,23],[65,97]])
print(var5)
print()
print(np.swapaxes(var5,0,1))
```

```python
# inverse matrix

print(np.linalg.inv(var5))
```

```python
# power of matrix
print(np.linalg.matrix_power(var5,4))
print()
print(np.linalg.matrix_power(var5,0))
print()
print(np.linalg.matrix_power(var5,-2))

# determinate of matrix

print(np.linalg.det(var5))
```
